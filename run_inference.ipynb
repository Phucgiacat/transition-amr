# IBM Transition AMR Parser - Batch Processing

This notebook runs the AMR parser with resume support.

**Features:**
- ✅ Resume after GPU disconnect
- ✅ Progress tracking (JSON file)
- ✅ Incremental output (safe writes)
- ✅ Configurable batch size

**Setup Steps:**
1. Run **Cell 2** once to create the Python 3.8 environment
2. Run **Cell 3** to start parsing with progress tracking

## Step 1: Setup Environment

This cell creates a Python 3.8 virtual environment and installs all dependencies.

**First run:** ~10-15 minutes (downloading PyTorch, fairseq, etc.)

**Subsequent runs:** Skip this step unless you need to reset the environment.

```python
import os
os.chdir('/content/transition-amr')
!bash setup.sh
```

## Step 2: Run Batch Inference

Parse sentences from `train.en(2).jsonl` and save AMR output to `train-amr.txt`.

**With progress tracking:**
- If interrupted, the next run will resume from where it stopped
- Progress is saved to `process.json`

**Without progress tracking:**
- Remove `--process` to run without checkpointing

```python
import os
os.chdir('/content/transition-amr')

!.venv/bin/python run.py \
    --input "train.en(2).jsonl" \
    --output "train-amr.txt" \
    --process "process.json" \
    --model "AMR3-structbart-L" \
    --batch-size 8
```

## Step 3: Check Progress (Optional)

View the current status of the parsing job.

```python
import os
os.chdir('/content/transition-amr')

!.venv/bin/python progress_utils.py --show process.json
```

## Step 4: Reset Progress (Optional)

If you want to start fresh and reparse all sentences, run this cell.

**Warning:** This backs up the current `process.json` and removes it, so the next inference run will start from the beginning.

```python
import os
os.chdir('/content/transition-amr')

!.venv/bin/python progress_utils.py --reset process.json
```

## Statistics

Count how many sentences have been parsed:

```python
import os
os.chdir('/content/transition-amr')

if os.path.exists('train-amr.txt'):
    with open('train-amr.txt', 'r', encoding='utf-8') as f:
        count = sum(1 for line in f if line.startswith('# ::id'))
    print(f"Parsed: {count} sentences")
else:
    print("train-amr.txt not found yet. Run the inference cell first.")
```
